````python
import os, io, json, time, contextlib, traceback, sys
from pathlib import Path
from typing import Dict, Any
from IPython.display import display, Markdown

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage

# --- Local packages ---
from mlkit import features, stats, cleaning, splits, metrics, reporting, schema_validator
from tools import profiler, research, hitl, model_design, logger
from prompts import NODE_PROMPT_TEMPLATES, SYSTEM_RULES

# --- Gemini setup ---
import getpass
if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Paste Gemini API key: ").strip()

LLM = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",  # Lite is fast & cheap; switch to flash if needed
    temperature=0.1,
    convert_system_message_to_human=True,
)

# --- Optional: research tool mode ---
RESEARCH_MODE = "auto"  # "tavily" | "gemini" | "stub" | "auto"
# If you have a Tavily key:
os.environ["TAVILY_API_KEY"] = ""

# --- Visibility flag ---
SHOW_CODE_INLINE = False  # True = show generated Python in notebook; False = only in log2_code.py

%matplotlib inline
sns.set_style("whitegrid")
print("✅ Environment ready")
````

````python
def get_inputs():
    print("-" * 50)
    print(" INPUTS - provide once at the start of every run")
    print("-" * 50)
    return {
        "train_path":         input("train_path (CSV/Parquet): ").strip(),
        "test_path":          input("test_path  (CSV/Parquet, optional - Enter to skip): ").strip(),
        "target_col":         input("target_col: ").strip(),
        "date_col":           input("date_col   (snapshot date, e.g. 'snapshot_date'): ").strip(),
        "business_objective": input("business_objective [Build CC propensity model]: ").strip() or "Build CC propensity model",
    }

USER = get_inputs()


# --- Set up artifact directories ---
from datetime import datetime
RUN_LOG = logger.RunLogger("artifacts/runs")
RUN_DIR = str(RUN_LOG.run_dir)

DIRS = {
    "root":   RUN_DIR,
    "eda":    f"{RUN_DIR}/eda",
    "clean":  f"{RUN_DIR}/clean",
    "design": f"{RUN_DIR}/design",
    "fe":     f"{RUN_DIR}/fe",
    "models": f"{RUN_DIR}/models",
    "eval":   f"{RUN_DIR}/eval",
    "logs":   RUN_DIR,  # logs live in run root
}

for p in DIRS.values():
    Path(p).mkdir(parents=True, exist_ok=True)
print(f"\n📂 Run directory: {RUN_DIR}")


# --- Load data ---
def _smart_load(path: str) -> pd.DataFrame:
    ext = Path(path).suffix.lower()
    if ext in (".parquet", ".pq"):
        return pd.read_parquet(path)
    # large CSV fallback to polars
    size_mb = Path(path).stat().st_size / 1e6 if Path(path).exists() else 0
    if size_mb > 500:
        try:
            import polars as pl
            return pl.read_csv(path, infer_schema_length=10000).to_pandas()
        except Exception:
            return pd.concat([c for c in pd.read_csv(path, chunksize=200_000, low_memory=False)], ignore_index=True)
    return pd.read_csv(path, low_memory=False)


df_train = _smart_load(USER["train_path"])
df_test  = _smart_load(USER["test_path"]) if USER["test_path"] else None

# --- Resolve target & date columns (typo-tolerant) ---
target_col = schema_validator.resolve_column(df_train, USER["target_col"])
date_col   = schema_validator.resolve_column(df_train, USER["date_col"])
print(f"\n🎯 Resolved target_col -> {target_col!r}")
print(f"📅 Resolved date_col   -> {date_col!r}")

# # Parse date column once
df_train[date_col] = pd.to_datetime(df_train[date_col], errors="coerce")
if df_test is not None:
    df_test[date_col] = pd.to_datetime(df_test[date_col], errors="coerce")

# --- Check train-test schema alignment ---
if df_test is not None:
    diff = schema_validator.diff_schemas(df_train, df_test)
    if diff["only_in_train"] or diff["only_in_test"]:
        print(f"\n⚠️ Schema mismatch - only_in_train: {diff['only_in_train'][:5]}, "
              f"only_in_test: {diff['only_in_test'][:5]}")

# --- Build the shared NS dict ---
NS: Dict[str, Any] = {
    "__builtins__":       __builtins__,
    "pd": pd, "np": np, "plt": plt, "sns": sns,
    "joblib": joblib, "json": json, "os": os,
    "df_train":           df_train,
    "df_test":            df_test,
    "target_col":         target_col,
    "date_col":           date_col,
    "business_objective": USER["business_objective"],
    "DIRS":               DIRS,
    "USER":               USER,
    "llm":                LLM,
    # mlkit & tools accessible from generated code
    "features": features, "stats": stats, "cleaning": cleaning,
    "splits": splits, "metrics": metrics, "reporting": reporting,
    "schema_validator": schema_validator,
}

print(f"\n🎯 NS initialized - train {df_train.shape}" +
      (f", test {df_test.shape}" if df_test is not None else " (no test set)"))
````

````python
def _build_stage_metadata(stage: str) -> dict:
    """Return what each stage reads/writes/assumes for log1 and log3."""
    META = {
        "profiling": {
            "inputs_from": ["user inputs (data_path, target_col, date_col)"],
            "outputs_to": ["model_design (consumes profile dict)"],
            "ns_keys_read": ["df_train", "target_col", "date_col", "business_objective"],
            "ns_keys_written": ["profile"],
            "assumptions": [
                "Pure observation - no decisions made here",
                "Only df_train profiled (df_test may be None)",
                "Date column parsed to datetime",
            ],
        },
        "model_design": {
            "inputs_from": ["profiling (NS['profile'])", "research_tool"],
            "outputs_to": ["all downstream stages (windows, algos, imputation)"],
            "ns_keys_read": ["profile", "business_objective", "llm"],
            "ns_keys_written": ["model_design", "research_findings"],
            "assumptions": [
                "OOT snapshots are the latest months (no future leakage)",
                "Algorithm shortlist constrained to 1-3 models",
                "Tier derived from profile's pos_ratio",
                "Validation: snapshots_for_training and oot_snapshots are disjoint",
            ],
        },
        "eda": {
            "inputs_from": ["profiling (NS['profile'])", "model_design (NS['model_design'])"],
            "outputs_to": ["cleaning, FE - eda_report is informational"],
            "ns_keys_read": [
                "df_train",
                "target_col",
                "date_col",
                "profile",
                "model_design.windows.snapshots_for_training",
            ],
            "ns_keys_written": ["eda_report"],
            "assumptions": [
                "EDA restricted to training snapshots to avoid OOT leakage",
                "Only top-15 correlated features visualized for clarity",
                "Histograms limited to top-3 by abs corr with target",
            ],
        },
        "cleaning": {
            "inputs_from": ["model_design (imputation & outlier strategies)"],
            "outputs_to": ["feature_engineering (cleaned df_train, df_test)"],
            "ns_keys_read": ["df_train", "df_test", "target_col", "model_design.data_strategy"],
            "ns_keys_written": [
                "df_train",
                "df_test",
                "cleaning_report",
                "imputation_map",
                "outlier_bounds",
            ],
            "assumptions": [
                "Imputation values learned from train only, applied to test (no leakage)",
                "Outlier bounds learned from train, clipped on both",
                "Duplicates dropped from train ONLY - test rows preserved",
                "Cols with >70% nulls dropped from both",
            ],
        },
        "feature_engineering": {
            "inputs_from": [
                "cleaning (df_train, df_test)",
                "model_design (windows, thresholds)",
            ],
            "outputs_to": ["training (X_train, y_train)", "evaluation (X_oot, y_oot)"],
            "ns_keys_read": [
                "df_train",
                "df_test",
                "target_col",
                "date_col",
                "llm",
                "model_design.windows",
                "model_design.data_strategy",
            ],
            "ns_keys_written": [
                "X_train",
                "X_oot",
                "y_train",
                "y_oot",
                "preprocessor",
                "feature_names",
                "column_map",
                "psi_report",
                "iv_report",
                "fe_report",
            ],
            "assumptions": [
                "Temporal split: OOT is FUTURE data - never random",
                "PSI threshold auto-drops unstable features (user can override via HITL)",
                "Correlation pruning keeps feature with higher target-correlation",
                "Scaling only applied if model_design requires it (e.g., for LogReg)",
                "Column mapping inferred by LLM with sample rows (overridable via HITL)",
            ],
        },
        "training": {
            "inputs_from": [
                "feature_engineering (X_train, y_train)",
                "model_design (algorithm shortlist, metric)",
            ],
            "outputs_to": ["evaluation (fitted_models)"],
            "ns_keys_read": [
                "X_train",
                "y_train",
                "model_design.algorithm_shortlist",
                "model_design.expected_metric_targets",
            ],
            "ns_keys_written": ["fitted_models", "cv_results", "imbalance_info"],
            "assumptions": [
                "Only algorithms from model_design shortlist trained (not full zoo)",
                "Imbalance handled via class_weight='balanced' or scale_pos_weight",
                "StratifiedKFold preserves class balance across CV folds",
                "Models verified to be distinct objects (no shared state)",
                "Score diversity check catches potential target leakage",
            ],
        },
        "evaluation": {
            "inputs_from": ["training (fitted_models)", "feature_engineering (X_oot, y_oot)"],
            "outputs_to": ["output (best_model, threshold, leaderboard)"],
            "ns_keys_read": [
                "X_oot",
                "y_oot",
                "fitted_models",
                "imbalance_info",
                "business_objective",
                "model_design.expected_metric_targets",
            ],
            "ns_keys_written": [
                "leaderboard",
                "best_model",
                "best_model_name",
                "optimal_threshold",
                "business_thresholds",
                "test_predictions",
                "lift_table",
                "metric_choice",
            ],
            "assumptions": [
                "OOT is the only true performance signal - CV scores not used for ranking",
                "Primary metric chosen by mlkit.metrics.select_primary_metric",
                "Threshold tuning: F1-optimal AND business-aware (precision >=0.5, recall >=0.8)",
                "Lift table computed for top-decile actionability (banking standard)",
            ],
        },
        "output": {
            "inputs_from": ["all upstream stages"],
            "outputs_to": ["disk (best_model.joblib, run_summary.json)"],
            "ns_keys_read": [
                "preprocessor",
                "best_model",
                "feature_names",
                "column_map",
                "model_design",
                "optimal_threshold",
                "imbalance_info",
            ],
            "ns_keys_written": ["artifact", "summary"],
            "assumptions": [
                "Artifact bundles preprocessor + model + thresholds + column_map for deployment",
                "summary JSON is the audit-trail snapshot for the run",
            ],
        },
    }
    return META.get(stage, {})


def _extract_decisions_from_ns(stage: str) -> dict:
    """Pull the key decisions made in this stage from NS for log1/log3."""
    if stage == "profiling":
        p = NS.get("profile", {})
        return {
            "n_rows": p.get("shape", {}).get("n_rows"),
            "n_cols": p.get("shape", {}).get("n_cols"),
            "target": p.get("target", {}).get("name"),
            "pos_ratio": p.get("target", {}).get("pos_ratio"),
            "n_snapshots": p.get("temporal", {}).get("n_snapshots"),
        }

    if stage == "model_design":
        md = NS.get("model_design", {})
        return {
            "algorithm_shortlist": md.get("algorithm_shortlist"),
            "primary_metric": md.get("expected_metric_targets", {}).get("primary_metric"),
            "imbalance_tier": md.get("data_strategy", {}).get("imbalance_tier"),
            "snapshots_for_training": md.get("windows", {}).get("snapshots_for_training"),
            "oot_snapshots": md.get("windows", {}).get("oot_snapshots"),
            "imputation_numeric": md.get("data_strategy", {}).get("imputation_numeric"),
            "outlier_strategy": md.get("data_strategy", {}).get("outlier_strategy"),
        }

    if stage == "eda":
        r = NS.get("eda_report", {})
        return {
            "snapshots_used": r.get("training_snapshots_used"),
            "n_top_correlations": len(r.get("top_correlations", [])),
            "eda_sample_rows": r.get("eda_sample_rows"),
        }

    if stage == "cleaning":
        r = NS.get("cleaning_report", {})
        return {
            "imputation_strategy": r.get("imputation_strategy"),
            "outlier_strategy": r.get("outlier_strategy"),
            "cols_dropped_high_null": r.get("dropped_columns", []),
            "duplicates_removed": r.get("train_duplicates_removed"),
            "rows_train_after": r.get("rows_train_after"),
        }

    if stage == "feature_engineering":
        r = NS.get("fe_report", {})
        return {
            "final_n_features": r.get("final_n_features"),
            "derived_created": r.get("created_derived_features"),
            "dropped_psi": r.get("dropped_psi_unstable"),
            "dropped_corr": len(r.get("dropped_correlation", [])),
            "woe_applied": r.get("woe_applied"),
            "scaling_applied": r.get("scaling_applied"),
        }

    if stage == "training":
        return {
            "models_trained": [r["model"] for r in NS.get("cv_results", [])],
            "imbalance_tier": NS.get("imbalance_info", {}).get("tier"),
            "cv_scores": {
                r["model"]: [v for k, v in r.items() if "mean" in k]
                for r in NS.get("cv_results", [])
            },
        }

    if stage == "evaluation":
        return {
            "best_model": NS.get("best_model_name"),
            "primary_metric": NS.get("metric_choice", {}).get("primary"),
            "optimal_threshold": NS.get("optimal_threshold"),
            "business_thresholds": NS.get("business_thresholds"),
        }

    if stage == "output":
        return {
            "artifact_path": f"{DIRS['models']}/best_model.joblib",
            "summary_path": f"{DIRS['root']}/run_summary.json",
        }
    return{}

# — original orchestrator helpers (with logging upgrades) —————————

def _safe_llm_invoke(prompt: str, max_net_retries: int = 3) -> str:
    """Resilient LLM call — handles transient network/quota errors."""
    for i in range(1, max_net_retries + 1):
        try:
            return LLM.invoke([HumanMessage(content=prompt)]).content
        except Exception as e:
            msg = str(e)
            transient = any(s in msg for s in [
                "getaddrinfo", "ConnectError", "RemoteDisconnected",
                "Read timed out", "ReadTimeout", "503", "504",
                "RESOURCE_EXHAUSTED", "429", "DEADLINE_EXCEEDED",
            ])
            if not transient or i == max_net_retries:
                if "API_KEY_INVALID" in msg or "API key not valid" in msg:
                    raise RuntimeError(
                        "❌ Gemini key invalid. Get a fresh one at "
                        "https://aistudio.google.com/app/apikey"
                    ) from e
                raise
            wait = 2 ** i
            print(f"🌐 Network hiccup ({msg[:60]}…) — retrying in {wait}s")
            time.sleep(wait)


def _extract_code(text: str) -> str:
    if "```" in text:
        for block in text.split("```"):
            b = block.strip()
            if b.startswith("python"):
                return b[6:].strip()
            if b.startswith("import") or "def " in b or "df_" in b or "NS[" in b:
                return b
    return text.strip()


def _run_code(code: str, ns: dict) -> str:
    ns["NS"] = ns  # make NS available as a key inside itself
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(code, ns)
    return buf.getvalue()


def run_llm_stage(stage_name: str, max_retries: int = 3,
                  extra_instructions: str = "") -> dict:
    """Run an LLM-driven stage with rich logging."""
    t0 = time.time()
    display(Markdown(f"## 🔵 Stage: **{stage_name.upper()}**"))
    base_prompt = NODE_PROMPT_TEMPLATES[stage_name]
    if extra_instructions:
        base_prompt += f"\n\nADDITIONAL USER INSTRUCTIONS:\n{extra_instructions}\n"
    
    last_err = ""
    for attempt in range(1, max_retries + 1):
        prompt = base_prompt + (
            f"\n\nPREVIOUS ATTEMPT ERROR — fix it:\n{last_err}" if last_err else ""
        )
        try:
            raw  = _safe_llm_invoke(prompt)
            code = _extract_code(raw)
        except Exception as e:
            display(Markdown(f"### ❌ LLM call failed: {e}"))
            raise

        RUN_LOG.log_code(stage_name, code, attempt)
        if SHOW_CODE_INLINE:
            display(Markdown(f"### 💻 Generated code (attempt {attempt})"))
            display(Markdown(f"```python\n{code}\n```"))

        try:
            out = _run_code(code, NS)
            if out.strip():
                display(Markdown(f"### 🟢 Output\n```\n{out[-4000:]}\n```"))
            plt.show()
            dur = round(time.time() - t0, 2)

            # — RICH LOGGING — log1 (cot) ————————————————
            stage_meta = _build_stage_metadata(stage_name)
            RUN_LOG.log_cot(
                stage=stage_name,
                summary=(f"Stage executed successfully on attempt {attempt} in {dur}s. "
                         f"Generated {len(code)} chars of Python code. "
                         f"Output produced {len(out)} chars of stdout."),
                assumptions=stage_meta.get("assumptions", []),
                decisions=_extract_decisions_from_ns(stage_name),
                inputs_from=stage_meta.get("inputs_from", []),
                outputs_to=stage_meta.get("outputs_to", []),
                ns_keys_read=stage_meta.get("ns_keys_read", []),
                ns_keys_written=stage_meta.get("ns_keys_written", []),
            )

            # — RICH LOGGING — log3 (tool I/O) ————————————
            RUN_LOG.log_tool(
                tool_name=stage_name,
                inputs={
                    "from_previous_stages": stage_meta.get("inputs_from", []),
                    "ns_keys_read":         stage_meta.get("ns_keys_read", []),
                    "prompt_chars":         len(prompt),
                    "extra_user_instructions": extra_instructions or None,
                    "attempt":              attempt,
                },
                outputs={
                    "key_decisions":   _extract_decisions_from_ns(stage_name),
                    "new_ns_keys":     stage_meta.get("ns_keys_written", []),
                    "stdout_preview":  out[:500],
                    "code_chars":      len(code),
                },
                duration_sec=dur,
                contract={
                    "reads_from_ns":  stage_meta.get("ns_keys_read", []),
                    "writes_to_ns":   stage_meta.get("ns_keys_written", []),
                    "depends_on":     stage_meta.get("inputs_from", []),
                },
            )
            display(Markdown(f"### ✅ {stage_name} completed (attempt {attempt}, {dur}s)"))
            return {"status": "ok", "attempt": attempt, "duration": dur}

        except Exception as e:
            last_err = traceback.format_exc()
            display(Markdown(f"### ⚠️ Attempt {attempt} failed: {e}"))
            display(Markdown(f"```\n{last_err[-1500:]}\n```"))

    RUN_LOG.log_tool(stage_name,
                     inputs={"prompt_len": len(base_prompt)},
                     outputs={}, errors=[last_err])
    raise RuntimeError(f"{stage_name} failed after {max_retries} attempts")


def run_with_hitl(
    stage_name: str,
    summary_builder,
    decision_builder=None,
    pre_stage_fn=None,
    max_iterations: int = 3,
):
    """Run a stage then ask HITL. Handle override/regenerate loops."""
    extra_instructions = ""
    for iteration in range(1, max_iterations + 1):
        if pre_stage_fn is not None:
            pre_stage_fn()
        else:
            run_llm_stage(stage_name, extra_instructions=extra_instructions)

        try:
            display(Markdown(summary_builder(NS)))
        except Exception as e:
            display(Markdown(f"⚠️ Summary failed: {e}"))

        if decision_builder is not None:
            decisions, options = decision_builder(NS)
        else:
            decisions, options = {}, {}

        hitl_result = hitl.hitl_gate(stage_name,
                                     summary=f"Stage `{stage_name}` complete.",
                                     proposed_decisions=decisions,
                                     options=options)

        # — RICH LOGGING — log3 records HITL with contract ————————
        RUN_LOG.log_tool(
            tool_name=f"hitl_{stage_name}",
            inputs={
                "decisions_shown": decisions,
                "options_shown":   options,
            },
            outputs={
                "user_action":  hitl_result["action"],
                "overrides":    hitl_result.get("overrides", {}),
                "instructions": hitl_result.get("instructions", ""),
            },
            hitl_decisions=[hitl_result],
            contract={
                "reads_from_ns":  [f"{stage_name} outputs"],
                "writes_to_ns":   ["model_design or NS keys via override"],
                "depends_on":     [f"{stage_name} (just completed)"],
            },
        )
        if hitl_result["action"] == "accept":
            return

        if hitl_result["action"] == "override":
            for k, v in hitl_result["overrides"].items():
                _apply_override(stage_name, k, v)
            return

        if hitl_result["action"] == "regenerate":
            extra_instructions = hitl_result["instructions"]
            print(f"\n🔄 Regenerating {stage_name} with user instructions...\n")
            continue

    print(f"⚠️ Reached max HITL iterations for {stage_name}; continuing with last result.")


def _apply_override(stage_name: str, key: str, value):
    """Apply a user override into NS."""
    if stage_name == "model_design":
        md = NS.get("model_design", {})
        if key in md:
            md[key] = value
        elif "data_strategy" in md and key in md["data_strategy"]:
            md["data_strategy"][key] = value
        elif "windows" in md and key in md["windows"]:
            md["windows"][key] = value
        elif key in ("algorithm_shortlist",):
            md[key] = value if isinstance(value, list) else [value]
        else:
            md[key] = value
        NS["model_design"] = md
        print(f"  → NS['model_design']['{key}'] = {value}")
        return

    if stage_name == "profiling":
        NS[key] = value
        print(f"  → NS['{key}'] = {value}")
        return

    NS[key] = value
    print(f"  → NS['{key}'] = {value}")

````

````python
def _profile_runner():
    t0 = time.time()
    print("🔍 Running profiler...")
    NS["profile"] = profiler.profile_dataset(
        df=NS["df_train"],
        target_col=NS["target_col"],
        date_col=NS["date_col"],
        business_objective=NS["business_objective"],
    )
    profiler.print_profile(NS["profile"])
    json.dump(NS["profile"], open(f"{DIRS['root']}/profile.json", "w"),
              indent=2, default=str)
    dur = round(time.time() - t0, 2)
    
    # Rich logging - log1 + log3
    stage_meta = _build_stage_metadata("profiling")
    RUN_LOG.log_cot(
        stage="profiling",
        summary=f"Profiled {NS['df_train'].shape[0]} rows & {NS['df_train'].shape[1]} cols in {dur}s",
        assumptions=stage_meta.get("assumptions", []),
        decisions=_extract_decisions_from_ns("profiling"),
        inputs_from=stage_meta.get("inputs_from", []),
        outputs_to=stage_meta.get("outputs_to", []),
        ns_keys_read=stage_meta.get("ns_keys_read", []),
        ns_keys_written=stage_meta.get("ns_keys_written", []),
    )
    RUN_LOG.log_tool(
        tool_name="profiling",
        inputs={
            "from_previous_stages": stage_meta.get("inputs_from", []),
            "ns_keys_read":         stage_meta.get("ns_keys_read", []),
            "df_shape":             list(NS["df_train"].shape),
        },
        outputs={
            "key_decisions": _extract_decisions_from_ns("profiling"),
            "new_ns_keys":   stage_meta.get("ns_keys_written", []),
        },
        duration_sec=dur,
        contract={
            "reads_from_ns": stage_meta.get("ns_keys_read", []),
            "writes_to_ns":  stage_meta.get("ns_keys_written", []),
            "depends_on":    stage_meta.get("inputs_from", []),
        },
    )
def _profile_summary(ns):
    p = ns["profile"]
    return (f"**Rows:** {p['shape']['n_rows']:,} | **Cols:** {p['shape']['n_cols']} | "
            f"**Target:** `{p['target']['name']}` "
            f"(pos_ratio={p['target'].get('pos_ratio')}) | "
            f"**Snapshots:** {p['temporal'].get('n_snapshots','?')}")

def _profile_decisions(ns):
    return {
        "target_col":         NS["target_col"],
        "date_col":           NS["date_col"],
        "business_objective": NS["business_objective"],
    }, {}

run_with_hitl("profiling", _profile_summary, _profile_decisions, pre_stage_fn=_profile_runner)
````

````python
def research(query: str, mode: str = "auto", llm=None, profile: dict = None) -> Dict:

    def _design_runner():
        t0 = time.time()
        print(f" Calling research tool (mode={RESEARCH_MODE})...")
        out = model_design.design_model(
            profile=NS["profile"],
            business_objective=NS["business_objective"],
            llm=llm,
            research_mode=RESEARCH_MODE,
        )

        NS["model_design"] = out["design"]
        NS["research_findings"] = out["research"]

        # FIX: use context managers to properly close file handles
        with open(f"{DIRS['design']}/model_design.json", "w") as f:
            json.dump(out["design"], f, indent=2, default=str)
        with open(f"{DIRS['design']}/research_findings.json", "w") as f:
            json.dump(out["research"], f, indent=2, default=str)

        dur = round(time.time() - t0, 2)

        # ==========================================
        # BUILD LOG1 REASONING - ALL from tool outputs, NOTHING hardcoded
        # ==========================================

        r = out["research"]        # research tool output
        md = out["design"]         # model design output
        p = NS["profile"]          # data profile

        snapshots = p.get("temporal", {}).get("snapshots", [])
        pos_ratio = p.get("target", {}).get("pos_ratio", "unknown")
        n_rows = p.get("shape", {}).get("n_rows", "?")
        n_cols = p.get("shape", {}).get("n_cols", "?")

        # --- Research section (everything from research tool) ---
        research_section = f"""### 🔍 Research Phase

**Research mode used:** `{r.get('mode', 'unknown')}`
**Query sent to research tool:** "{r.get('query', 'n/a')}"
**Use case matched:** {r.get('matched_use_case', 'n/a')}
**Detected from data:** pos_ratio={r.get('detected_pos_ratio', 'n/a')} -> tier={r.get('detected_tier', 'n/a')}

**Research summary:**
{r.get('summary', 'No summary provided by research tool')}

---

**Algorithms recommended by research tool:** {r.get('recommended_algorithms', 'n/a')}

**Why these algorithms? (reasoning from research tool):**
{r.get('algorithm_reasoning', 'No algorithm reasoning provided by research tool')}

---

**Windows recommended by research tool:** {r.get('windows', 'n/a')}

**Why these windows? (reasoning from research tool):**
{r.get('window_reasoning', 'No window reasoning provided by research tool')}

---

**Metrics recommended by research tool:** {r.get('metrics', 'n/a')}

**Why these metrics? (reasoning from research tool):**
{r.get('metric_reasoning', 'No metric reasoning provided by research tool')}

---

**Class weight strategy (from research):** {r.get('class_weight_strategy', 'n/a')}
**SMOTE recommended (from research):** {r.get('smote_recommended', 'n/a')}
**Key features suggested:** {r.get('key_features', 'n/a')}
**Research sources:** {r.get('sources', ['No sources'])}
"""

        # --- Model Design section (everything from LLM) ---
        design_section = f"""### 🏗️ Model Design Decisions (LLM decisions based on research + data profile)

**Problem framing:** {md.get('problem_framing', 'n/a')}

**Data context that informed decisions:**
- Dataset: {n_rows} rows x {n_cols} cols
- Positive rate: {pos_ratio}
- Available snapshots: {snapshots} ({len(snapshots)} months)

---

#### Snapshot allocation

**Training snapshots:** {md.get('windows', {}).get('snapshots_for_training', [])}
**OOT snapshots:** {md.get('windows', {}).get('oot_snapshots', [])}

**Why this OOT split? (LLM reasoning):**
{md.get('windows', {}).get('oot_rationale', 'No OOT rationale provided by LLM')}

**Why this observation/performance window? (LLM reasoning):**
{md.get('windows', {}).get('window_rationale', md.get('rationale', 'No window rationale provided by LLM'))}

---

#### Algorithm selection

**Algorithms selected:** {md.get('algorithm_shortlist', [])}

**Why these algorithms? (LLM reasoning):**
{md.get('algorithm_rationale', md.get('rationale', 'No algorithm rationale provided by LLM'))}

---

#### Metric selection

**Primary metric:** {md.get('expected_metric_targets', {}).get('primary_metric', '?')}

**Why this metric? (LLM reasoning):**
{md.get('metric_rationale', 'No metric rationale provided by LLM')}

---

#### Data strategy

| Decision | Value |
|---|---|
| Imputation (numeric) | {md.get('data_strategy', {}).get('imputation_numeric', '?')} |
| Imputation (categorical) | {md.get('data_strategy', {}).get('imputation_categorical', '?')} |
| Outlier strategy | {md.get('data_strategy', {}).get('outlier_strategy', '?')} |
| Correlation threshold | {md.get('data_strategy', {}).get('correlation_threshold', '?')} |
| PSI threshold | {md.get('data_strategy', {}).get('psi_threshold', '?')} |
| Imbalance tier | {md.get('data_strategy', {}).get('imbalance_tier', '?')} |
| Scaling required | {md.get('data_strategy', {}).get('scaling_required', '?')} |
| WoE transformation | {md.get('data_strategy', {}).get('woe_transformation', '?')} |

**Why this strategy? (LLM reasoning):**
{md.get('data_strategy', {}).get('strategy_rationale', md.get('rationale', 'No strategy rationale provided by LLM'))}

---

#### Overall rationale (from LLM)
{md.get('rationale', 'No overall rationale provided')}
"""

        # --- Errors section ---
        errors_section = ""
        if out.get("errors"):
            errors_section = "\n#### ⚠️ Design process notes:\n"
            for e in out["errors"]:
                errors_section += f"- {e}\n"
            if "FALLBACK_USED" in str(out.get("errors")):
                errors_section += (
                    "\n**FALLBACK DESIGN USED** - LLM failed to produce valid JSON "
                    "after 3 attempts. Deterministic defaults were applied based on "
                    "data profile and research findings.\n"
                )

        # ==========================================
        # WRITE TO LOG1 + LOG3
        # ==========================================

        stage_meta = _build_stage_metadata("model_design")

        full_summary = f"Model Design completed in {dur}s.\n\n{research_section}\n{design_section}\n{errors_section}"

        # FIX: was log_cct (typo) — corrected to log_cot
        RUN_LOG.log_cot(
            stage="model_design",
            summary=full_summary,
            assumptions=stage_meta.get("assumptions", []),
            decisions=_extract_decisions_from_ns("model_design"),
            inputs_from=stage_meta.get("inputs_from", []),
            outputs_to=stage_meta.get("outputs_to", []),
            ns_keys_read=stage_meta.get("ns_keys_read", []),
            ns_keys_written=stage_meta.get("ns_keys_written", []),
        )

        RUN_LOG.log_tool(
            tool_name="model_design",
            inputs={
                "from_previous_stages": stage_meta.get("inputs_from", []),
                "ns_keys_read":         stage_meta.get("ns_keys_read", []),
                "business_objective":   NS["business_objective"],
                "research_mode":        RESEARCH_MODE,
            },
            outputs={
                "key_decisions":       _extract_decisions_from_ns("model_design"),
                "new_ns_keys":         stage_meta.get("ns_keys_written", []),
                "research_mode_used":  r.get("mode"),
                "detected_tier":       r.get("detected_tier"),
                "research_sources":    r.get("sources", [])[:5],
            },
            duration_sec=dur,
            contract={
                "reads_from_ns": stage_meta.get("ns_keys_read", []),
                "writes_to_ns":  stage_meta.get("ns_keys_written", []),
                "depends_on":    stage_meta.get("inputs_from", []),
            },
            errors=out.get("errors", []),
        )

        # Pretty-print to notebook
        print("\n" + "=" * 60)
        print("MODEL DESIGN DECISIONS")
        print("=" * 60)
        print(f"Problem framing : {md.get('problem_framing')}")
        print(f"\nWindows:")
        for k, v in md.get("windows", {}).items():
            print(f"  {k:30s}: {v}")
        print(f"\nData Strategy:")
        for k, v in md.get("data_strategy", {}).items():
            print(f"  {k:30s}: {v}")
        print(f"\nAlgorithm shortlist: {md.get('algorithm_shortlist')}")
        print(f"Primary metric      : {md.get('expected_metric_targets', {}).get('primary_metric')}")
        print(f"\nRationale: {md.get('rationale')}")
        print("=" * 60)

        # FIX: added explicit return so caller gets a consistent value
        return {
            "status": "ok",
            "duration": dur,
            "design": md,
            "research": r,
        }

    # FIX: _design_summary, _design_decisions, and run_with_hitl are now
    # correctly placed OUTSIDE _design_runner but INSIDE research()
    def _design_summary(ns):
        md = ns["model_design"]
        return (
            f"**Algorithms:** {md.get('algorithm_shortlist')} | "
            f"**Primary metric:** {md.get('expected_metric_targets', {}).get('primary_metric')} | "
            f"**Tier:** {md.get('data_strategy', {}).get('imbalance_tier')} | "
            f"**OOT:** {md.get('windows', {}).get('oot_snapshots')}"
        )

    def _design_decisions(ns):
        md = ns["model_design"]
        decisions = {
            "algorithm_shortlist":    md.get("algorithm_shortlist"),
            "imputation_numeric":     md.get("data_strategy", {}).get("imputation_numeric"),
            "outlier_strategy":       md.get("data_strategy", {}).get("outlier_strategy"),
            "correlation_threshold":  md.get("data_strategy", {}).get("correlation_threshold"),
            "psi_threshold":          md.get("data_strategy", {}).get("psi_threshold"),
            "snapshots_for_training": md.get("windows", {}).get("snapshots_for_training"),
            "oot_snapshots":          md.get("windows", {}).get("oot_snapshots"),
        }
        options = {
            "algorithm_shortlist":   "LogisticRegression,RandomForest,XGBoost,LightGBM,GradientBoosting",
            "imputation_numeric":    "median,mean,zero",
            "outlier_strategy":      "winsorize_99,iqr_safe,none",
            "correlation_threshold": "0.80,0.85,0.90,0.95",
            "psi_threshold":         "0.10,0.20,0.25,0.30",
        }
        return decisions, options

    run_with_hitl("model_design", _design_summary, _design_decisions, pre_stage_fn=_design_runner)
````

````python
def _eda_summary(ns):
    r = ns.get("eda_report", {})
    return (f"**EDA sample rows:** {r.get('eda_sample_rows','?')}; | "
            f"**Snapshots used:** {r.get('training_snapshots_used')} | "
            f"**Top correlations** computed, **missingness** mapped.")

def _eda_decisions(ns):
    return {"reviewed_correlations": True, "reviewed_distributions": True}, {}

run_with_hitl("eda", _eda_summary, _eda_decisions)



````

````python
def _clean_summary(ns):
    r = ns.get("cleaning_report", {})
    return (f"**Train:** {r.get('rows_train_before','?')}->{r.get('rows_train_after','?')} | "
            f"**Nulls cleared:** {r.get('nulls_train_before','?')}->{r.get('nulls_train_after',0)} | "
            f"**Outlier strategy:** {r.get('outlier_strategy')}")

def _clean_decisions(ns):
    ds = ns.get("model_design", {}).get("data_strategy", {})
    return {
        "imputation_numeric":    ds.get("imputation_numeric"),
        "imputation_categorical": ds.get("imputation_categorical"),
        "outlier_strategy":      ds.get("outlier_strategy"),
    }, {
        "imputation_numeric":    "median,mean,zero",
        "imputation_categorical": "mode,constant",
        "outlier_strategy":      "winsorize_99,iqr_safe,none",
    }

run_with_hitl("cleaning", _clean_summary, _clean_decisions)
````

````python
def _fe_summary(ns):
    r = ns.get("fe_report", {})
    return (f"**Derived features:** {len(r.get('created_derived_features', []))} | "
            f"**Final features:** {r.get('final_n_features','?')} | "
            f"**Dropped (PSI):** {len(r.get('dropped_psi_unstable', []))} | "
            f"**Dropped (corr):** {len(r.get('dropped_correlation', []))} | "
            f"**WoE applied:** {r.get('woe_applied', False)}")

def _fe_decisions(ns):
    ds = ns.get("model_design", {}).get("data_strategy", {})
    return {
        "correlation_threshold": ds.get("correlation_threshold", 0.85),
        "psi_threshold":         ds.get("psi_threshold", 0.25),
        "scaling_required":      ds.get("scaling_required", False),
        "woe_transformation":    ds.get("woe_transformation", False),
        "column_map":            ns.get("column_map", {}),
    }, {
        "correlation_threshold": "0.80,0.85,0.90,0.95",
        "psi_threshold":         "0.10,0.20,0.25,0.30",
    }

run_with_hitl("feature_engineering", _fe_summary, _fe_decisions)
````

````python
def _train_summary(ns):
    cvr = ns.get("cv_results", [])
    imb = ns.get("imbalance_info", {})
    rows = "\n".join(
        f"  - **{r['model']}**: {r.get('scoring')} = "
        f"{[v for k,v in r.items() if 'mean' in k][0]:.4f}"
        for r in cvr
    )
    return (f"**Imbalance tier:** {imb.get('tier')} | "
            f"**Models trained:** {len(cvr)}\n\n{rows}")

def _train_decisions(ns):
    return {
        "algorithm_shortlist": ns.get("model_design", {}).get("algorithm_shortlist"),
        "scoring":             ns.get("cv_results", [{}])[0].get("scoring"),
    }, {}

run_with_hitl("training", _train_summary, _train_decisions)


````

````python
def _eval_summary(ns):
    lb = ns.get("leaderboard")
    best = ns.get("best_model_name")
    mc = ns.get("metric_choice", {})
    if lb is None or not best:
        return "Evaluation pending..."
    pm = mc.get("primary", "F1")
    val = lb.iloc[0].get(pm, None)
    base = (f"**Best model:** `{best}` | **Primary metric ({pm}):** {val:.4f}"
            if val is not None else f"**Best:** `{best}` | metric n/a")
    return f"{base} | **Threshold:** {ns.get('optimal_threshold', 0.5):.3f}"

def _eval_decisions(ns):
    bt = ns.get("business_thresholds", {})
    return {
        "best_model":        ns.get("best_model_name"),
        "optimal_threshold": ns.get("optimal_threshold"),
        "threshold_f1":      bt.get("f1_optimal"),
        "threshold_p50":     bt.get("precision_50"),
        "threshold_r80":     bt.get("recall_80"),
    }, {}

run_with_hitl("evaluation", _eval_summary, _eval_decisions)
````

````python
def _output_summary(ns):
    return (f"**Run summary saved** at `{DIRS['root']}/run_summary.json` | "
            f"**Best model artifact** at `{DIRS['models']}/best_model.joblib`")

def _output_decisions(ns):
    return {}, {}

run_with_hitl("output", _output_summary, _output_decisions)


````
